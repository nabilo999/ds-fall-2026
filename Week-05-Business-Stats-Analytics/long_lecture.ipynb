## Load the data

````python
import pandas as pd
import numpy as np
import plotly.express as px

from data_profiling import ProfileReport
````

````python
df_un = pd.read_csv('zenodo-data/unsponsored_posts.csv')
df_un['sponsor_type'] = 'Unsponsored'
print(df_un.shape)
df_un.head()
````

````python
df_sponsored = pd.read_csv('zenodo-data/sponsored_posts.csv')
df_sponsored['sponsor_type'] = 'Sponsored'
print(df_sponsored.shape)
df_sponsored.head()
````

## HOW BIG IS THE DATA? 

````python
df_un.shape, df_sponsored.shape
````

## Cleaning the data

````python
# Lets join them together. 

## Okay, lets join them into one dataframe, 
## and then we can do some analysis on the combined dataset.
df = pd.concat([df_un, df_sponsored], ignore_index=True, sort=False)

# Mention the index trap, and how to avoid it. # (appending/concatenating two dfs will defaul index duplicate indicies)  # dont spend any time on it tho, you can avoid by includeing the ignore_index=True parameter in the concat function, which we did above.
print(df_un.shape, df_sponsored.shape, df.shape)
````

````python
# CHECK FOR DUPES
df.duplicated().sum()
````

````python
# CHECK FOR NULLS
df.isnull().sum()

# Is null just meaning 0 or does 0 mean 0 and null means None?
````

````python
# Looking at just the rows that have null values.

# There are 364 Nulls in user_followers.  Let's see what those rows look like.
df[df.isnull()['user_followers']]
````

````python
# CHecking to see if there are posts that have 0 and not Null in them
# is NaN the same as 0?
df[df.user_followers < 1]
````

````python
# sanity check, seeing if a null id is present the the 0 values.  it is not. lets drop and continue. 
id_with_nan = 2298140344686880783
df_nulls = df[df.user_followers < 1]

df_nulls[df_nulls['id'] == id_with_nan]
````

````python
# no they are different, lets just drop them as the seem like mistakes and less than 1% of the data. 
print(df.shape)
df = df.dropna(subset=['user_followers'])
df.shape
````

## Box plot showing the Quartiles.  
25% of users live in each box.  

````python
fig = px.box(
    unique_users_df,
    y="user_followers",
    points='all',
    title="Distribution of Instagram User Followers, RAW log_y=False",
)

fig.show()



fig = px.box(
    unique_users_df,
    y="user_followers",
    points='all',
    title="Distribution of Instagram User Followers, log_y=True",
    log_y=True
)
fig.show()



unique_users_df["followers_log"] = np.log10(unique_users_df["user_followers"] + 1)

fig = px.box(
    unique_users_df,
    y="followers_log",
    points="all",
    title="Instagram Follower Counts (log10 + 1)"
)

fig.show()


fig = px.box(
    unique_users_df,
    y="user_followers",
    points=False,
    title="Instagram Follower Counts, RAW, not logged just limiting Rage [0,10000]"
)

fig.update_yaxes(range=[0, 10000])

fig.show()
````
